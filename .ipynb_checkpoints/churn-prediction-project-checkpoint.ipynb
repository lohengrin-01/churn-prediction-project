### 1. Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('dataset.csv')
df.head()
```

```python
df.head().T
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
categorical_columns = list(df.dtypes[df.dtypes == 'object'].index)
for c in categorical_columns:
    df[c] = df[c].str.lower().str.replace(' ', '_')
```

```python
df.head().T
```

```python
df.dtypes
```

```python
df.totalcharges = pd.to_numeric(df.totalcharges, errors='coerce')
df.totalcharges = df.totalcharges.fillna(0)
```

```python
df.churn
```

```python
df.churn = (df.churn == 'Yes').astype(int)
```

```python
df.churn.head()
```

### 2. Setting up the validation framework

```python
from sklearn.model_selection import train_test_split
```

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
```

```python
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
y_train = df_train.churn.values
y_val = df_val.churn.values
y_test = df_test.churn.values
```

```python
del df_train['churn']
del df_val['churn']
del df_test['churn']
```

### 3. EDA

```python
df_full_train.isnull().sum()
```

### 4. Feature importance

```python
global_churn
```

```python
for c in categorical:
    print(c)
    df_group = df_full_train.groupby(c).churn.agg(['mean', 'count'])
    df_group['diff'] = df_group['mean'] - global_churn
    df_group['risk'] = df_group['mean'] / global_churn
    display(df_group)
    print()
    print()
```

### 5. One-Hot Encoding

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import OneHotEncoder
```

```python
X_train_num = df_train[numerical].values

scaler = StandardScaler()
#scaler = MinMaxScaler()

X_train_num = scaler.fit_transform(X_train_num)
```

```python
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
X_train_cat = ohe.fit_transform(df_train[categorical].values)
```

```python
X_train = np.column_stack([X_train_num, X_train_cat])
```

```python
X_val_num = scaler.transform(df_val[numerical].values)
```

```python
X_val_cat = ohe.transform(df_val[categorical].values)
```

```python
X_val = np.column_stack([X_val_num, X_val_cat])
```

### 6. Model

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
model.coef_[0].round(3)
```

```python
y_pred = model.predict_proba(X_val)[:, 1]
```

```python
churn_decision = (y_pred >= 0.5)
```

```python
df_pred = pd.DataFrame()
df_pred['probability'] = y_pred
df_pred['prediction'] = churn_decision.astype(int)
df_pred['actual'] = y_val
df_pred['correct'] = df_pred['prediction'] == df_pred['actual']
```

```python
df_pred
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_val, y_pred >= 0.5)
```

### 9. Accuracy

```python
thresholds = np.linspace(0, 1, 21)
scores = []
for t in thresholds:
    score = accuracy_score(y_test, y_pred >= t)
    print('%2f %3f' % (t, score))
    scores.append(score)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression

# Datos originales, antes del escalado y del encoding
X = df_train[numerical + categorical]
y = y_train

# Preparación de las columnas
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numerical),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical)
])

# Preparación + modelo
pipeline = make_pipeline(
    preprocessor,
    LogisticRegression(max_iter=1000)
)

# Definir las cinco particiones
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Entrenar y evaluar en cada partición
scores = cross_val_score(
    pipeline, X, y,
    cv=cv,
    scoring="roc_auc"
)

print("AUC de cada vuelta:", scores)
print(f"AUC promedio: {scores.mean():.3f}")
print(f"Desviación estándar: {scores.std():.3f}")
```
